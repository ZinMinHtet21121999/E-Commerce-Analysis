```python
def data_check(df,col):
    df[col] = pd.to_numeric(df[col],errors='coerce')
```

```python
def name_check(df,col):
    df[col] = df[col].str.strip().str.title()
    df[col] = df[col].fillna('Unknown')
```

```python
def fill_with_groupby(df,gro,col):
    df[col] = df.groupby(gro)[col].transform(lambda x : x.fillna(x.median()))
```

```python
def date_clean(df,col):
    df[col] = pd.to_datetime(df[col])
    df[col] = df[col].fillna(method='bfill').fillna(method='ffill')
```

```python
data_check(df,'OrderID')
```

```python
data_check(df,'CustomerID')
```

```python
name_check(df,'ProductCategory')
name_check(df,'ProductName')
name_check(df,'Region')
```

```python
fill_with_groupby(df,'ProductName','Quantity')
fill_with_groupby(df,'ProductName','UnitPrice')
```

```python
date_clean(df,'OrderDate')
```
